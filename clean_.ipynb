```python
import re
```

```python
def clean_job_title(title):
    if pd.isna(title):
        return title
    
    s = str(title)
    
    # 1. Bỏ số + dấu phẩy ở đầu
    s = re.sub(r'^\d+,\s*', '', s)
    
    # 2. Bỏ tất cả nội dung trong () và []
    s = re.sub(r'\(.*?\)', '', s)
    s = re.sub(r'\[.*?\]', '', s)
    
    # 3. Bỏ các từ khóa tuyển dụng dư thừa
    s = re.sub(r'Tuyển Dụng|TUYỂN DỤNG', '', s, flags=re.IGNORECASE)
    
    # 4. Bỏ ký tự _ hoặc - dư
    s = re.sub(r'[_\-]+', ' ', s)
    
    # 5. Chuẩn hóa khoảng trắng
    s = re.sub(r'\s+', ' ', s)
    
    return s.strip()
```

```python
dfs = {
    "df_1" : df_1,
    "df_2" : df_2,
    "df_3" : df_3,
    "df_4" : df_4
}
for name, df in dfs.items():
    if 'job_title' in df.columns:
        df['job_title'] = df['job_title'].apply(clean_job_title)
        df.dropna(subset=['job_title'], inplace=True)
```

```python
def format_date(date):
    if pd.isna(date):
        return pd.NaT

    date = str(date)

    # Book3 – ISO 8601
    if 'T' in date:
        return pd.to_datetime(date, errors='coerce').date()

    # Book1, Book2 – dd/mm/yyyy
    return pd.to_datetime(date, format="%d/%m/%Y", errors='coerce').date()
```

```python
for name, df in dfs.items():
    if 'date' in df.columns:
        df['date'] = df['date'].apply(format_date)
        df.dropna(subset=['date'], inplace = True)
```

```python
from deep_translator import GoogleTranslator

translator = GoogleTranslator(source='vi', target='en')

def safe_translate(text):
    if not isinstance(text, str) or text.strip() == "":
        return text
    try:
        return translator.translate(text)
    except:
        return text   # giữ nguyên nếu không dịch được
```

```python
import numpy as np
```

```python
def salary_to_million(salary):
    if pd.isna(salary):
        return np.nan
    
    s = str(salary).lower()
    
    if any(x in s for x in ['thỏa thuận', 'thương lượng', 'cạnh tranh']):
        return np.nan
    
    nums = re.findall(r'\d+[.,]?\d*', s)
    nums = [float(n.replace(',', '')) for n in nums]
    
    if not nums:
        return np.nan
    
    max_value = max(nums)
    
    if 'triệu' in s or 'tr' in s:
        return max_value
    
    if 'usd' in s or '$' in s:
        return round((max_value * 27000) / 1_000_000, 2)
    
    return np.nan
```

```python
for name, df in dfs.items():
    if 'salary' not in df.columns:
        continue
    
    df['salary_million'] = df['salary'].apply(salary_to_million)

    df['mean_city_job'] = (
        df
        .groupby(['location', 'job_title'])['salary_million']
        .transform('mean')
    )

    df['mean_city'] = (
        df
        .groupby('location')['salary_million']
        .transform('mean')
    )

    global_mean = df['salary_million'].mean()

    df['salary'] = (
        df['salary_million']
        .fillna(df['mean_city_job'])
        .fillna(df['mean_city'])
        .fillna(global_mean)
    )

    df.drop(
        columns=['salary_million', 'mean_city_job', 'mean_city'],
        inplace=True
    )
```

```python
def location_clean(location):
    if pd.isna(location):
        return np.nan
    
    s = str(location)
    
    # 1. Bỏ nội dung trong ngoặc ()
    s = re.sub(r'\(.*?\)', '', s)
    
    # 2. Nếu có nhiều địa điểm, lấy cái đầu tiên
    s = s.split(',')[0]
    
    # 3. Chuẩn hóa khoảng trắng
    s = re.sub(r'\s+', ' ', s)
    
    return s.strip()
```

```python
for name, df in dfs.items():
    if 'location' in df.columns:
        df['location'] = df['location'].apply(location_clean)
        df.dropna(subset=['location'], inplace = True)
```

```python
def clean_company_name(company):
    if pd.isna(company):
        return company

    s = str(company)

    # 1. Bỏ số + dấu phẩy ở đầu (vd: "123, Công ty ABC")
    s = re.sub(r'^\d+,\s*', '', s)

    # 2. Bỏ địa điểm trong () hoặc []
    s = re.sub(
        r'\((Hà Nội|HCM|TP\.? HCM|TPHCM|Đà Nẵng|Vietnam|Việt Nam).*?\)',
        '',
        s,
        flags=re.IGNORECASE
    )
    s = re.sub(
        r'\[(Hà Nội|HCM|TP\.? HCM|Đà Nẵng|Vietnam|Việt Nam).*?\]',
        '',
        s,
        flags=re.IGNORECASE
    )

    # 3. Bỏ từ khóa dư thừa liên quan tuyển dụng
    s = re.sub(
        r'Tuyển Dụng|TUYỂN DỤNG|Hiring|Recruitment',
        '',
        s,
        flags=re.IGNORECASE
    )

    # 4. Chuẩn hóa loại hình công ty
    s = re.sub(
        r'\b(Công ty TNHH|TNHH|Cổ phần|CP|JSC|.,JSC|Co\.,? Ltd|Company Limited|Ltd\.?)\b',
        '',
        s,
        flags=re.IGNORECASE
    )

    # 5. Bỏ ký tự thừa
    s = re.sub(r'[_\-]+', ' ', s)

    # 6. Chuẩn hóa khoảng trắng
    s = re.sub(r'\s+', ' ', s)

    return s.strip()
```

```python
for name, df in dfs.items():
    if 'company' in df.columns:
        df['company'] = df['company'].apply(clean_company_name)
        df.dropna(subset=['company'], inplace = True)
```
